The author shows that functions like `ROW_NUMBER`, `RANK`, `LAG`, and `LEAD` make it possible to rank rows, compare values, and compute running totals while keeping all detail rows intact.
```sql
/* ============================================================
   TITLE: Running Total of Sales Per Salesperson Over Time
   WHY SPECIAL:
   - Uses window aggregate SUM() with PARTITION BY + ORDER BY
   - Keeps detail rows while adding a running total (can't do with GROUP BY alone)
   - Classic Chapter 7 window-frame example
   ============================================================ */

SELECT
    soh.SalesPersonID,
    soh.OrderDate,
    soh.SalesOrderID,
    soh.TotalDue,
    SUM(soh.TotalDue) OVER(
        PARTITION BY soh.SalesPersonID
        ORDER BY soh.OrderDate, soh.SalesOrderID
        ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
    ) AS RunningTotal
FROM Sales.SalesOrderHeader AS soh
WHERE soh.SalesPersonID IS NOT NULL
ORDER BY soh.SalesPersonID, soh.OrderDate, soh.SalesOrderID;
```

```sql
/* ============================================================
   TITLE: Order Value Compared to Customer Average (Window AVG)
   WHY SPECIAL:
   - Uses AVG() OVER to compute customer-level averages
   - Keeps all detail rows and adds a comparison column
   - Classic analytic query comparing row vs. partition statistic
   ============================================================ */

SELECT
    soh.CustomerID,
    soh.SalesOrderID,
    soh.TotalDue AS OrderValue,
    AVG(soh.TotalDue) OVER(
        PARTITION BY soh.CustomerID
    ) AS CustomerAvgOrderValue,
    soh.TotalDue - AVG(soh.TotalDue) OVER(
        PARTITION BY soh.CustomerID
    ) AS DifferenceFromAverage
FROM Sales.SalesOrderHeader AS soh
ORDER BY soh.CustomerID, soh.SalesOrderID;
```

```sql
/* ============================================================
   TITLE: Top 3 Orders Per Customer (ROW_NUMBER)
   WHY SPECIAL:
   - ROW_NUMBER ranks within each partition (cust)
   - Lets you filter to "top N per group" in outer query
   - Ranking window function core topic in Chapter 7
   ============================================================ */

WITH C AS (
    SELECT
        soh.CustomerID,
        soh.SalesOrderID,
        soh.TotalDue,
        ROW_NUMBER() OVER(
            PARTITION BY soh.CustomerID
            ORDER BY soh.TotalDue DESC, soh.SalesOrderID DESC
        ) AS rn
    FROM Sales.SalesOrderHeader AS soh
)
SELECT CustomerID, SalesOrderID, TotalDue
FROM C
WHERE rn <= 3
ORDER BY CustomerID, TotalDue DESC;
```

```sql
/* ============================================================
   TITLE: Price Deciles for Products (NTILE)
   WHY SPECIAL:
   - NTILE groups rows into equal-sized buckets
   - Useful for distribution / segmentation analysis
   - Demonstrates ranking + analytic binning
   ============================================================ */

SELECT
    p.ProductID,
    p.Name,
    p.ListPrice,
    NTILE(10) OVER(ORDER BY p.ListPrice) AS PriceDecile
FROM Production.Product AS p
WHERE p.ListPrice > 0
ORDER BY p.ListPrice;
```

```sql
/* ============================================================
   TITLE: Previous/Next Order Value Per Customer (LAG/LEAD)
   WHY SPECIAL:
   - Offset window functions look backward/forward without self-joins
   - Great for change-over-time calculations
   - Chapter 7 offset functions section
   ============================================================ */

SELECT
    soh.CustomerID,
    soh.SalesOrderID,
    soh.OrderDate,
    soh.TotalDue,
    LAG(soh.TotalDue, 1) OVER(
        PARTITION BY soh.CustomerID
        ORDER BY soh.OrderDate, soh.SalesOrderID
    ) AS PrevOrderValue,
    LEAD(soh.TotalDue, 1) OVER(
        PARTITION BY soh.CustomerID
        ORDER BY soh.OrderDate, soh.SalesOrderID
    ) AS NextOrderValue
FROM Sales.SalesOrderHeader AS soh
ORDER BY soh.CustomerID, soh.OrderDate, soh.SalesOrderID;
```

```sql
/* ============================================================
   TITLE: First and Last Order Date Per Customer
   WHY SPECIAL:
   - FIRST_VALUE/LAST_VALUE need explicit frames to be correct
   - Shows "value from start/end of partition" per row
   - Chapter 7 FIRST_VALUE/LAST_VALUE example
   ============================================================ */

SELECT
    soh.CustomerID,
    soh.SalesOrderID,
    soh.OrderDate,
    FIRST_VALUE(soh.OrderDate) OVER(
        PARTITION BY soh.CustomerID
        ORDER BY soh.OrderDate, soh.SalesOrderID
        ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
    ) AS FirstOrderDate,
    LAST_VALUE(soh.OrderDate) OVER(
        PARTITION BY soh.CustomerID
        ORDER BY soh.OrderDate, soh.SalesOrderID
        ROWS BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING
    ) AS LastOrderDate
FROM Sales.SalesOrderHeader AS soh
ORDER BY soh.CustomerID, soh.OrderDate, soh.SalesOrderID;
```

```sql
/* ============================================================
   TITLE: Pivot Total Sales by Territory and Year (PIVOT)
   WHY SPECIAL:
   - Rotates rows -> columns inside SQL (pivot table behavior)
   - Demonstrates Chapter 7 pivoting with PIVOT operator
   - Cleaner than many CASE aggregates
   ============================================================ */

SELECT TerritoryID, [2012], [2013], [2014]
FROM (
    SELECT
        soh.TerritoryID,
        YEAR(soh.OrderDate) AS OrderYear,
        soh.TotalDue
    FROM Sales.SalesOrderHeader AS soh
    WHERE soh.TerritoryID IS NOT NULL
) AS D
PIVOT(
    SUM(TotalDue) FOR OrderYear IN ([2012],[2013],[2014])
) AS P
ORDER BY TerritoryID;
```

```sql
/* ============================================================
   TITLE: Pivot Same Result Using GROUP BY + CASE
   WHY SPECIAL:
   - Standard (portable) pivot technique
   - Makes grouping/spreading/aggregation phases explicit
   - Chapter 7 "pivoting with grouped query" method
   ============================================================ */

SELECT
    soh.TerritoryID,
    SUM(CASE WHEN YEAR(soh.OrderDate) = 2012 THEN soh.TotalDue END) AS [2012],
    SUM(CASE WHEN YEAR(soh.OrderDate) = 2013 THEN soh.TotalDue END) AS [2013],
    SUM(CASE WHEN YEAR(soh.OrderDate) = 2014 THEN soh.TotalDue END) AS [2014]
FROM Sales.SalesOrderHeader AS soh
WHERE soh.TerritoryID IS NOT NULL
GROUP BY soh.TerritoryID
ORDER BY soh.TerritoryID;
```

```sql
/* ============================================================
   TITLE: Grouping Sets for Territory + Year + Grand Total
   WHY SPECIAL:
   - Multiple grouping levels in ONE scan
   - Replaces long UNION ALL patterns
   - Uses GROUPING_ID to label which grouping set produced each row
   ============================================================ */

SELECT
    GROUPING_ID(soh.TerritoryID, YEAR(soh.OrderDate)) AS GroupingSetID,
    soh.TerritoryID,
    YEAR(soh.OrderDate) AS OrderYear,
    SUM(soh.TotalDue) AS TotalSales
FROM Sales.SalesOrderHeader AS soh
GROUP BY GROUPING SETS(
    (soh.TerritoryID, YEAR(soh.OrderDate)),
    (soh.TerritoryID),
    (YEAR(soh.OrderDate)),
    ()
)
ORDER BY GroupingSetID, soh.TerritoryID, OrderYear;
```

```sql
/* ============================================================
   TITLE: Monthly Sales Buckets + Gap Filling (Time Series)
   WHY SPECIAL:
   - Buckets time series using DATEADD/DATEDIFF (works in SQL 2019)
   - Generates missing months and left-joins to "fill gaps"
   - Mirrors Chapter 7 time-series + gap-fill idea without SQL 2022 funcs
   ============================================================ */

DECLARE
    @startdate DATE = '2012-01-01',
    @enddate   DATE = '2014-12-01';

WITH Months AS (
    -- recursive month generator
    SELECT @startdate AS MonthStart
    UNION ALL
    SELECT DATEADD(month, 1, MonthStart)
    FROM Months
    WHERE MonthStart < @enddate
),
Bucketed AS (
    SELECT
        DATEADD(month, DATEDIFF(month, 0, soh.OrderDate), 0) AS MonthStart,
        SUM(soh.TotalDue) AS SalesInMonth
    FROM Sales.SalesOrderHeader AS soh
    WHERE soh.OrderDate >= @startdate
      AND soh.OrderDate <  DATEADD(month, 1, @enddate)
    GROUP BY DATEADD(month, DATEDIFF(month, 0, soh.OrderDate), 0)
)
SELECT
    m.MonthStart,
    b.SalesInMonth
FROM Months AS m
LEFT OUTER JOIN Bucketed AS b
    ON m.MonthStart = b.MonthStart
ORDER BY m.MonthStart
OPTION (MAXRECURSION 1000);
```
